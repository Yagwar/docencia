# INFERENCIA MEDIANTE INTERVALOS DE CONFIANZA (95%)

## INTERVALO DE CONFIANZA PARA UNA MEDIA POBLACIONAL (Mu)
Variable: 'Depresion_Beck' (Sintomatología depresiva BDI-II, escala 0 a 63 pts)

```r
# Paso A: Extraer insumos de la muestra
n <- length(datos$Depresion_Beck)                 # Tamaño muestral (n = 200)
media_x <- mean(datos$Depresion_Beck)            # Media muestral (x_bar = 18.40)
desv_s <- sd(datos$Depresion_Beck)               # Desviación estándar (s = 7.80)
gl <- n - 1                                       # Grados de libertad (df = 199)

# Paso B: Calcular el Error Estándar de la Media (SE)
se_media <- desv_s / sqrt(n)                     # SE = s / sqrt(n) = 0.5515

# Paso C: Obtener el valor crítico t de Student (Confianza = 0.95, alpha = 0.05)
t_critico <- qt(0.975, df = gl)                  # t_critico = 1.972

# Paso D: Calcular los límites del intervalo de confianza
limite_inf_media <- media_x - (t_critico * se_media)
limite_sup_media <- media_x + (t_critico * se_media)

# Imprimir cálculo manual
cat("\n=== 1.1. IC AL 95% PARA LA MEDIA DE DEPRESIÓN DE BECK ===\n")
cat(sprintf("Media Muestral: %.2f | Error Estándar: %.4f\n", media_x, se_media))
cat(sprintf("IC 95%% Manual: [ %.2f , %.2f ]\n", limite_inf_media, limite_sup_media))

# Paso E: Verificación automática en R
test_media_ic <- t.test(datos$Depresion_Beck, conf.level = 0.95)
cat("Verificación automática con t.test():\n")
print(test_media_ic$conf.int)
```

## INTERVALO DE CONFIANZA PARA UNA PROPORCIÓN POBLACIONAL (P)
Variable: 'Vapeo_Semanal' (Categoría de interés: "Sí")

```r
# Paso A: Extraer conteo de éxitos y proporción muestral
exitos_vapeo <- sum(datos$Vapeo_Semanal == "Sí")   # Conteo de consumidores (x = 71)
p_hat <- exitos_vapeo / n                         # Proporción muestral (p_hat = 0.355)

# Paso B: Calcular el Error Estándar de la Proporción (SE_p)
se_prop <- sqrt((p_hat * (1 - p_hat)) / n)        # SE = sqrt(p_hat*(1-p_hat)/n) = 0.0338

# Paso C: Obtener el valor crítico Z (Confianza = 0.95, alpha = 0.05)
z_critico <- qnorm(0.975)                         # Z_critico = 1.96

# Paso D: Calcular los límites del intervalo de confianza
limite_inf_prop <- p_hat - (z_critico * se_prop)
limite_sup_prop <- p_hat + (z_critico * se_prop)

# Imprimir cálculo manual
cat("\n=== 1.2. IC AL 95% PARA LA PROPORCIÓN DE VAPEO SEMANAL ===\n")
cat(sprintf("Proporción Muestral (p_hat): %.3f (%.1f%%) | SE: %.4f\n", p_hat, p_hat*100, se_prop))
cat(sprintf("IC 95%% Manual: [ %.4f , %.4f ] (%.2f%% a %.2f%%)\n", 
            limite_inf_prop, limite_sup_prop, limite_inf_prop*100, limite_sup_prop*100))

# Paso E: Verificación automática en R
test_prop_ic <- prop.test(x = exitos_vapeo, n = n, conf.level = 0.95, correct = FALSE)
cat("Verificación automática con prop.test():\n")
print(test_prop_ic$conf.int)
```

# PRUEBAS DE HIPÓTESIS PARA UNA MUESTRA

##  PRUEBA DE HIPÓTESIS PARA UNA MEDIA (t / Z)

CASO 1 (SIGNIFICATIVO): 'Horas_Sueno' frente a la norma médica (Mu_0 = 7.0 h)
H0: Mu >= 7.0 (Los estudiantes duermen igual o más de 7 horas)
H1: Mu < 7.0  (Los estudiantes duermen menos de 7 horas - Unilateral Izquierda)
Alpha = 0.05

```r
mu_0_sueno <- 7.0
media_sueno <- mean(datos$Horas_Sueno)
se_sueno <- sd(datos$Horas_Sueno) / sqrt(n)

# Estadístico t calculado
t_calc_sueno <- (media_sueno - mu_0_sueno) / se_sueno

# p-valor (área acumulada a la izquierda de t_calc)
p_val_sueno <- pt(t_calc_sueno, df = gl)

cat("\n=== 2.1A. PRUEBA DE HIPÓTESIS MEDIA: SUEÑO (CASO SIGNIFICATIVO) ===\n")
cat(sprintf("Media Observada: %.2f h | Mu_0: %.1f h | t: %.4f | p-valor: %.10f\n", 
            media_sueno, mu_0_sueno, t_calc_sueno, p_val_sueno))
cat("Decisión: RECHAZAR H0 (Existe significancia estadística, p < 0.05)\n")

# Verificación con t.test()
t.test(datos$Horas_Sueno, mu = 7.0, alternative = "less")
```

CASO 2 (NO SIGNIFICATIVO): 'Puntaje_Cognitivo' frente a la norma de prueba (Mu_0 = 100 pts)

H0: Mu = 100.0 (El puntaje promedio es igual a la norma)

H1: Mu != 100.0 (El puntaje promedio difiere de la norma - Bilateral)

Alpha = 0.05

```r
mu_0_cog <- 100.0
media_cog <- mean(datos$Puntaje_Cognitivo)
se_cog <- sd(datos$Puntaje_Cognitivo) / sqrt(n)

# Estadístico t calculado
t_calc_cog <- (media_cog - mu_0_cog) / se_cog

# p-valor bilateral (dos colas)
p_val_cog <- 2 * (1 - pt(abs(t_calc_cog), df = gl))

cat("\n=== 2.1B. PRUEBA DE HIPÓTESIS MEDIA: COGNICIÓN (CASO NO SIGNIFICATIVO) ===\n")
cat(sprintf("Media Observada: %.2f pts | Mu_0: %.1f pts | t: %.4f | p-valor: %.4f\n", 
            media_cog, mu_0_cog, t_calc_cog, p_val_cog))
cat("Decisión: NO RECHAZAR H0 (No hay significancia estadística, p >= 0.05)\n")

# Verificación con t.test()
t.test(datos$Puntaje_Cognitivo, mu = 100.0, alternative = "two.sided")
```

## PRUEBA DE HIPÓTESIS PARA UNA PROPORCIÓN (Z)

CASO 1 (SIGNIFICATIVO): 'Vapeo_Semanal' frente a tasa nacional (p_0 = 0.25)
H0: p <= 0.25
H1: p > 0.25 (Unilateral Derecha)

```r
p_0_vapeo <- 0.25
se_0_vapeo <- sqrt((p_0_vapeo * (1 - p_0_vapeo)) / n)   # SE bajo H0
z_calc_vapeo <- (p_hat - p_0_vapeo) / se_0_vapeo
p_val_vapeo <- 1 - pnorm(z_calc_vapeo)

cat("\n=== 2.2A. PRUEBA DE HIPÓTESIS PROPORCIÓN: VAPEO (CASO SIGNIFICATIVO) ===\n")
cat(sprintf("p_hat: %.3f | p_0: %.2f | Z: %.4f | p-valor: %.6f\n", 
            p_hat, p_0_vapeo, z_calc_vapeo, p_val_vapeo))
cat("Decisión: RECHAZAR H0 (La prevalencia supera significativamente el 25%)\n")

# Verificación con prop.test()
prop.test(x = exitos_vapeo, n = n, p = 0.25, alternative = "greater", correct = FALSE)
```

CASO 2 (NO SIGNIFICATIVO): 'Asistencia_Terapia' frente a referente histórico (p_0 = 0.14)
H0: p = 0.14
H1: p != 0.14 (Bilateral)

```r
exitos_terapia <- sum(datos$Asistencia_Terapia == "Sí")
p_hat_terapia <- exitos_terapia / n
p_0_terapia <- 0.14
se_0_terapia <- sqrt((p_0_terapia * (1 - p_0_terapia)) / n)
z_calc_terapia <- (p_hat_terapia - p_0_terapia) / se_0_terapia
p_val_terapia <- 2 * (1 - pnorm(abs(z_calc_terapia)))

cat("\n=== 2.2B. PRUEBA DE HIPÓTESIS PROPORCIÓN: TERAPIA (CASO NO SIGNIFICATIVO) ===\n")
cat(sprintf("p_hat: %.3f | p_0: %.2f | Z: %.4f | p-valor: %.4f\n", 
            p_hat_terapia, p_0_terapia, z_calc_terapia, p_val_terapia))
cat("Decisión: NO RECHAZAR H0 (No hay diferencia significativa respecto al 14%)\n")

# Verificación con prop.test()
prop.test(x = exitos_terapia, n = n, p = 0.14, alternative = "two.sided", correct = FALSE)
```

# PRUEBA DE HIPÓTESIS PARA DIFERENCIA DE MEDIAS

## DIFERENCIA DE MEDIAS EN MUESTRES INDEPENDIENTES (Student vs Welch)

Comparar 'Depresion_Beck' entre estudiantes que vapean vs que no vapean

H0: Mu_Vapeo - Mu_NoVapeo = 0  (No hay diferencia en depresión)

H1: Mu_Vapeo - Mu_NoVapeo != 0 (Existe diferencia en depresión)

```r
# Paso A: Dividir subgrupos y calcular estadísticos descriptivos
grupo_vapeo <- datos$Depresion_Beck[datos$Vapeo_Semanal == "Sí"]
grupo_novapeo <- datos$Depresion_Beck[datos$Vapeo_Semanal == "No"]

n1 <- length(grupo_vapeo)                          # n1 = 71
n2 <- length(grupo_novapeo)                        # n2 = 129

m1 <- mean(grupo_vapeo)                            # Media grupo 1
m2 <- mean(grupo_novapeo)                          # Media grupo 2

s1_sq <- var(grupo_vapeo)                          # Varianza grupo 1
s2_sq <- var(grupo_novapeo)                        # Varianza grupo 2
```

```r
# Paso B: Evaluación preliminar de Homogeneidad de Varianzas (Prueba F)
f_var_test <- var.test(Depresion_Beck ~ Vapeo_Semanal, data = datos)

cat("\n=== 3.1. PRUEBA t PARA DOS MUESTRAS INDEPENDIENTES (DEPRESIÓN vs VAPEO) ===\n")
cat(sprintf("Grupo Vapeo (Sí): n1 = %d, Media1 = %.2f, DE1 = %.2f\n", n1, m1, sqrt(s1_sq)))
cat(sprintf("Grupo No Vapeo:   n2 = %d, Media2 = %.2f, DE2 = %.2f\n", n2, m2, sqrt(s2_sq)))
cat(sprintf("Prueba F de Homogeneidad de Varianzas p-valor: %.4f\n", f_var_test$p.value))
```

```r
# Paso C: Cálculo de Error Estándar y Estadístico t (Caso Varianzas Iguales / Pooled)
sp_sq <- ((n1 - 1) * s1_sq + (n2 - 1) * s2_sq) / (n1 + n2 - 2)   # Varianza mancomunada
se_diff_pooled <- sqrt(sp_sq * ((1 / n1) + (1 / n2)))            # SE de la diferencia
gl_pooled <- n1 + n2 - 2                                        # df = 198

t_calc_indep <- (m1 - m2) / se_diff_pooled
p_val_indep <- 2 * (1 - pt(abs(t_calc_indep), df = gl_pooled))

cat(sprintf("Cálculo Manual t (Pooled): t = %.4f | df = %d | p-valor = %.6f\n", 
            t_calc_indep, gl_pooled, p_val_indep))
```

```r
# Paso D: Verificación automática en R con t.test()
cat("\nVerificación con t.test() en R (Varianzas Iguales, var.equal = TRUE):\n")
print(t.test(Depresion_Beck ~ Vapeo_Semanal, data = datos, var.equal = TRUE))

cat("\nVerificación con t.test() en R (Ajuste de Welch por defecto, var.equal = FALSE):\n")
print(t.test(Depresion_Beck ~ Vapeo_Semanal, data = datos, var.equal = FALSE))
```

## DIFERENCIA DE MEDIAS EN MUESTRAS APAREADAS / RELACIONADAS

Escenario clínico: Evaluación de intervención breve en 'Estrés/Ansiedad' (Pre vs Post)

Se simula una medición repetida sobre los mismos N = 200 estudiantes.

```r
set.seed(2026)
ansiedad_pre <- round(rnorm(n, mean = 28.5, sd = 5.2), 1)
ansiedad_post <- round(ansiedad_pre - rnorm(n, mean = 4.2, sd = 2.1), 1)
```

```r
# Paso A: Calcular la variable de diferencias individuales (d = Post - Pre)
diferencias <- ansiedad_post - ansiedad_pre
```

```r
# Paso B: Estadísticos descriptivos de la diferencia
d_bar <- mean(diferencias)                         # Media de las diferencias
s_d <- sd(diferencias)                             # Desviación de las diferencias
se_d <- s_d / sqrt(n)                              # SE de las diferencias
gl_paired <- n - 1
```

```r
# Paso C: Cálculo del estadístico t apareado
t_calc_paired <- (d_bar - 0) / se_d
p_val_paired <- pt(t_calc_paired, df = gl_paired)  # Unilateral (post < pre)

cat("\n=== 3.2. PRUEBA t PARA MUESTRAS APAREADAS (INTERVENCIÓN EN ANSIEDAD) ===\n")
cat(sprintf("Media Pre: %.2f | Media Post: %.2f | Media Diferencia (d_bar): %.2f\n", 
            mean(ansiedad_pre), mean(ansiedad_post), d_bar))
cat(sprintf("Cálculo Manual: t = %.4f | df = %d | p-valor = %.10f\n", 
            t_calc_paired, gl_paired, p_val_paired))
cat("Decisión: RECHAZAR H0 (La intervención redujo significativamente la ansiedad)\n")
```

```r
# Paso D: Verificación automática en R
cat("\nVerificación con t.test(paired = TRUE):\n")
print(t.test(ansiedad_post, ansiedad_pre, paired = TRUE, alternative = "less"))
```
