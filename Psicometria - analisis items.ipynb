```r
# if (!require(psych)) install.packages("psych")
library(psych)
```

```r
# Cargar la base de datos desde el archivo CSV
datos <- read.csv("datos_simulados_psicometria.csv")
```

# RECODIFICACIÓN DE ÍTEMS INVERTIDOS / NEGATIVOS (Ítems 21 a 25)
Escala Likert de 1 a 6 -> Fórmula de inversión: (Mínimo + Máximo) - X = 7 - X



```r
items_invertidos <- paste0("Item_", 21:25)
datos_recodificados <- datos

for (item in items_invertidos) {
  datos_recodificados[[item]] <- 7 - datos_recodificados[[item]]
}
```

# PROPIEDADES DESCRIPTIVAS POR ÍTEM (TENDENCIA CENTRAL Y DISPERSIÓN)

```r
tabla_descriptiva <- data.frame(
  Item = colnames(datos_recodificados),
  Media = round(colMeans(datos_recodificados), 2),
  Mediana = round(apply(datos_recodificados, 2, median), 2),
  DE = round(apply(datos_recodificados, 2, sd), 2),
  Varianza = round(apply(datos_recodificados, 2, var), 2),
  Min = apply(datos_recodificados, 2, min),
  Max = apply(datos_recodificados, 2, max)
)

cat("--- MEDIDAS DE TENDENCIA CENTRAL Y DISPERSIÓN POR ÍTEM ---\n")
print(tabla_descriptiva)
```

# CORRELACIONES INTER-ÍTEM

```r
matriz_cor <- round(cor(datos_recodificados), 3)
```

```r
cat("\n--- MATRIZ DE CORRELACIÓN INTER-ÍTEM (Primeros 5 ítems) ---\n")
print(matriz_cor[1:5, 1:5])
```

```r
promedio_cor_inter_item <- mean(matriz_cor[lower.tri(matriz_cor)])
cat("\nCorrelación Promedio Inter-Ítem:", round(promedio_cor_inter_item, 3), "\n")
```

# ANÁLISIS DE CONFIABILIDAD Y DISCRIMINACIÓN CON 'psych'
La función psych::alpha() estima la consistencia interna, las correlaciones ítem-total (bruta r y corregida r.drop) y el Alfa si el ítem es eliminado (alpha.drop).



```r
analisis_psych <- psych::alpha(datos_recodificados)

cat("\n======================================================\n")
cat("RESULTADOS GLOBALES DE CONFIABILIDAD (psych::alpha)\n")
cat("======================================================\n")
print(round(analisis_psych$total, 4))
```

```r
# Construcción de la tabla integral de análisis de ítems
tabla_psicometrica <- data.frame(
  Item = rownames(analisis_psych$item.stats),
  Media = round(analisis_psych$item.stats$mean, 2),
  DE = round(analisis_psych$item.stats$sd, 2),
  r_item_total = round(analisis_psych$item.stats$raw.r, 3),      
  r_it_corregida = round(analisis_psych$item.stats$r.drop, 3),  
  Alfa_si_elimina = round(analisis_psych$alpha.drop$raw_alpha, 4)
)

cat("\n--- PROPIEDADES PSICOMÉTRICAS Y DISCRIMINACIÓN DE ÍTEMS ---\n")
print(tabla_psicometrica)
```

## DEPURACIÓN DE LA PRUEBA (Criterio de discriminación r_it_corregida >= 0.20)


```r
items_conservados <- tabla_psicometrica$Item[tabla_psicometrica$r_it_corregida >= 0.20]
items_eliminados <- tabla_psicometrica$Item[tabla_psicometrica$r_it_corregida < 0.20]

datos_finales <- datos_recodificados[, items_conservados]
analisis_final <- psych::alpha(datos_finales)

cat("\n======================================================\n")
cat("RESULTADO DE LA DEPURACIÓN DE LA ESCALA\n")
cat("======================================================\n")
cat("Ítems Retenidos (", length(items_conservados), "de 25):", paste(items_conservados, collapse = ", "), "\n")
cat("Ítems Eliminados:", paste(items_eliminados, collapse = ", "), "\n")
cat("Alfa de Cronbach Final (Escala Depurada):", round(analisis_final$total$raw_alpha, 4), "\n")
```

 # ESTANDARIZACIÓN Y TABLA GENERAL DE BAREMOS

```r
X_directo <- rowSums(datos_finales)
media_X <- mean(X_directo)
sd_X <- sd(X_directo)
```

```r
# Generación de la tabla normativa por puntaje bruto posible en el rango observado
puntajes_unicos <- sort(unique(X_directo))

baremo <- data.frame(
  Puntaje_Directo = puntajes_unicos,
  Frecuencia = as.vector(table(X_directo)),
  Percentil = round(ecdf(X_directo)(puntajes_unicos) * 100, 1)
)
```

```r
# Puntuaciones transformadas lineales
baremo$Puntaje_Z <- round((baremo$Puntaje_Directo - media_X) / sd_X, 2)
baremo$Puntaje_T <- round(50 + (10 * baremo$Puntaje_Z), 1)

# Puntuación T Normalizada (a partir del percentil acumulado)
p_adj <- pmin(pmax(baremo$Percentil / 100, 0.001), 0.999) # Evita infinitos en colas
baremo$Puntaje_T_Normalizado <- round(50 + 10 * qnorm(p_adj), 1)
```

```r
cat("======================================================\n")
cat("TABLA GENERAL DE BAREMOS NORMATIVOS (PRUEBA FINAL)\n")
cat("======================================================\n")
print(baremo)
```
