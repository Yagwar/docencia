```r
set.seed(20262)
n <- 250
```

```r
# A. Variables Demográficas y de Identidad
genero <- sample(c("Femenino", "Masculino"), size = n, replace = TRUE, prob = c(0.55, 0.45))
localidad <- sample(c("Suba", "Kennedy", "Chapinero", "Usaquén", "Fontibón", "Otro"), 
                    size = n, replace = TRUE, prob = c(0.25, 0.20, 0.15, 0.15, 0.15, 0.10))
estrato <- sample(1:6, size = n, replace = TRUE, prob = c(0.03, 0.12, 0.44, 0.31, 0.07, 0.03))

# Edad en años cumplidos (Promedio ~20 años, nacidos ~2006)
edad <- round(pmax(pmin(rnorm(n, mean = 20.1, sd = 1.4), 26), 17))

# B. Variables de Riesgo y Conducta Digital (Generación Z en 2026)
vapeo_semanal <- sample(c("Sí", "No"), size = n, replace = TRUE, prob = c(0.38, 0.62))
frecuencia_alcohol <- sample(c("Nunca", "Ocasional", "Frecuentemente"), size = n, replace = TRUE, prob = c(0.28, 0.68, 0.04))
red_preferida <- sample(c("TikTok", "Instagram", "Discord", "YouTube"), 
                        size = n, replace = TRUE, prob = c(0.40, 0.38, 0.15, 0.07))

# Tiempo diario en pantalla (Horas) - Escalar de Razón (Asimetría Positiva / Log-Normal)
tiempo_pantalla <- rlnorm(n, meanlog = 1.9, sdlog = 0.4)
tiempo_pantalla <- round(pmax(pmin(tiempo_pantalla, 16.0), 1.5), 1)

# C. Variables Psicológicas y Medición
# Bienestar Subjetivo - Escalar de Intervalo (Simétrica / Normal, 0-100 pts)
bienestar <- rnorm(n, mean = 68.0, sd = 12.0)
bienestar <- round(pmax(pmin(bienestar, 100.0), 0.0), 1)

# Soledad Percibida - Escalar de Intervalo (0-10 pts)
soledad_percibida <- round(pmax(pmin(rnorm(n, mean = 4.5, sd = 2.4), 10), 0))

# Ansiedad sin celular (Nomofobia) - Variable Ordinal
ansiedad_celular <- sample(c("1. Muy baja", "2. Baja", "3. Moderada", "4. Alta", "5. Muy alta"), 
                           size = n, replace = TRUE, prob = c(0.05, 0.15, 0.25, 0.35, 0.20))
```

```r
datos_estudiantes <- data.frame(
  ID = sprintf("EST_%03d", 1:n),
  Edad = edad,
  Genero = factor(genero),
  Localidad = factor(localidad),
  Estrato = factor(estrato, ordered = TRUE),
  Vapeo_Semanal = factor(vapeo_semanal),
  Frecuencia_Alcohol = factor(frecuencia_alcohol, levels = c("Nunca", "Ocasional", "Frecuentemente"), ordered = TRUE),
  Red_Preferida = factor(red_preferida),
  Tiempo_Pantalla = tiempo_pantalla,
  Bienestar_Subjetivo = bienestar,
  Soledad_Percibida = soledad_percibida,
  Ansiedad_Sin_Celular = factor(ansiedad_celular, ordered = TRUE)
)
```

```r
write.csv(datos_estudiantes, "datos_estudiantes_bogota.csv", row.names = FALSE)
```
