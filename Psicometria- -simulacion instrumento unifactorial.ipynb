```r
set.seed(42) # Semilla ajustada para lograr Alpha inicial ~ 63%
N <- 150
n_items <- 25
```

# SIMULACIÓN DE DATOS (MODELO UNIFACTORIAL)

```r
# Habilidad o rasgo latente de los estudiantes
theta <- rnorm(N, mean = 0, sd = 1)

# Cargas factoriales (discriminación subyacente ajustada)
# Ítems 1-12: Buenos (0.28 - 0.38)
# Ítems 13-15: Sesgados con sesgo de respuesta/piso/techo (0.25 - 0.30)
# Ítems 16-20: Débiles/Ruido (0.01 - 0.12)
# Ítems 21-25: Invertidos (-0.22 a -0.32)
loadings <- c(
  c(0.32, 0.35, 0.30, 0.38, 0.34, 0.29, 0.31, 0.36, 0.33, 0.28, 0.37, 0.30),
  c(0.28, 0.25, 0.27),
  c(0.05, 0.02, 0.11, 0.08, -0.03),
  c(-0.25, -0.30, -0.22, -0.28, -0.32)
)

# Sesgos/Efectos piso y techo (Shifts de intercepto)
shifts <- c(
  rep(0, 12),
  -1.8, 1.8, -1.5, # Ítems sesgados (efecto piso/techo)
  rep(0, 10)
)

# Matriz de respuestas continuas: Y_ij = Shift_j + Loading_j * Theta_i + e_ij
Y <- matrix(0, nrow = N, ncol = n_items)
for (j in 1:n_items) {
  e <- rnorm(N, mean = 0, sd = 1)
  Y[, j] <- shifts[j] + loadings[j] * theta + e
}

# Categorización a Escala Likert de 6 puntos (1 a 6)
datos_mat <- matrix(0, nrow = N, ncol = n_items)
cortes <- c(-1.1, -0.4, 0.2, 0.8, 1.4)

for (j in 1:n_items) {
  cat_index <- findInterval(Y[, j], cortes) + 1
  datos_mat[, j] <- pmin(pmax(cat_index, 1), 6)
}

datos <- as.data.frame(datos_mat)
colnames(datos) <- paste0("Item_", 1:n_items)
```

```r
write.csv(datos, file = "datos_simulados_psicometria.csv", row.names = FALSE)
```
